```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Date': pd.date_range(start="2020-01-01", periods=10, freq='M'),
    'Close': [1350, 1380, 1420, 1450, 1500, 1550, 1600, 1580, 1620, 1650],
    'Volume': [1200000, 1100000, 1300000, 1250000, 1400000, 1350000, 1500000, 1450000, 1550000, 1600000]
}

df = pd.DataFrame(data)

start_date = "2020-03-01"
end_date = "2020-09-30"
mask = (df['Date'] >= start_date) & (df['Date'] <= end_date)
filtered = df.loc[mask]

plt.figure(figsize=(10,5))
plt.plot(filtered['Date'], filtered['Close'], marker='o', linestyle='-', color='blue')
plt.title("Alphabet Inc. Stock Prices (Mar–Sep 2020)")
plt.xlabel("Date")
plt.ylabel("Closing Price")
plt.grid(True)
plt.show()

```

```python
plt.figure(figsize=(10,5))
plt.bar(filtered['Date'], filtered['Volume'])
plt.title("Alphabet Inc. Trading Volume")
plt.xlabel("Date")
plt.ylabel("Volume")
plt.xticks(rotation=45)
plt.show()

```

```python
plt.figure(figsize=(8,6))
plt.scatter(filtered['Volume'], filtered['Close'], alpha=0.7)
plt.title("Alphabet Inc. Volume vs Stock Price")
plt.xlabel("Trading Volume")
plt.ylabel("Stock Price")
plt.show()

```

```python
import numpy as np

df_random = pd.DataFrame(np.random.randn(10,4), columns=list('ABCD'))

def highlight_negative(val):
    color = 'red' if val < 0 else 'black'
    return f'color: {color}'
styled = df_random.style.applymap(highlight_negative)
styled

```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame(np.random.randn(10,4), columns=list('ABCD'))
df.iloc[2,1] = np.nan
df.iloc[5,3] = np.nan
df.iloc[7,0] = np.nan
def highlight_nan(val):
    color = 'red' if pd.isna(val) else 'black'
    return f'color: {color}'

styled = df.style.applymap(highlight_nan)
styled

```
